```python
import matplotlib.pyplot as plt
```

```python
plt.subplot(2,2,2)
plt.hist(df["age"])
```

```python
plt.hist(df["physical_activity"])
plt.title("Physical activity")
plt.show()
```

```python
plt.hist(df["sleep_hours"])
plt.title("sleep hours")
plt.show()
```

```python
sns.countplot(x="platform_usage", data=df)
plt.title("Platform Usage")
plt.show()
```

```python
sns.countplot(x="social_interaction_level", data=df)
plt.title("social_interaction_level")
plt.show()
```

```python

sns.heatmap(df.corr(numeric_only=True), annot=True)
plt.title("Heatmap")
plt.show()
```

```python
gender = df["gender"].value_counts()
plt.pie(gender, labels=gender.index, autopct="%1.1f%%")
plt.title("Gender Distribution")
plt.show()
```

```python
gender = df["age"].value_counts()
plt.pie(gender, labels=gender.index, autopct="%1.1f%%")
plt.title("age")
plt.show()
```

```python
gender = df["daily_social_media_hours"].value_counts()
plt.pie(gender, labels=gender.index, autopct="%1.1f%%")
plt.title("daily_social_media_hours")
plt.show()
```

```python
plt.scatter(df["addiction_level"], df["depression_label"])
plt.title("addiction level vs depression labele")
plt.xlabel("addiction_level")
plt.ylabel("depression_label")
plt.show()
```

```python
plt.scatter(df["daily_social_media_hours"], df["academic_performance"])
plt.title("daily_social_media_hours vs academic_performance")
plt.xlabel("daily_social_media_hours")
plt.ylabel("academic_performance")
plt.show()
```

```python
sns.boxplot(y=df["daily_social_media_hours"])

plt.title("Box Plot of daily social media hours")
plt.show()
```

```python
sns.boxplot(y=df["academic_performance"])

plt.title("Box Plot of academic performance")
plt.show()
```

```python
plt.plot(df["age"], df["gender"])
plt.title("Age vs gender")
plt.xlabel("Age")
plt.ylabel("gender")
plt.show()
```

```python
plt.plot(df["anxiety_level"], df["stress_level"])
plt.title("anxietylevel vs stress_level")
plt.xlabel("anxiety_level")
plt.ylabel("stress_level")
plt.show()
```
